```python
import pandas as pd
import numpy as np
```

```python
data= pd.read_csv("/content/drive/MyDrive/MACHINE LEARNING/Student Awareness Survey (Responses) - Form Responses 1.csv")
```

```python
data.head(5)
```

```python
data = data.drop(columns=['Timestamp','Email','Job role that you are interested in','What is the minimum salary of students placed through campus (In LPA..respond as a number)','What is the maximum salary of students placed through campus (In LPA..respond as a number)','What is the median salary of students placed through campus (In LPA..respond as a number)','Which is the highest paying company that recruits from campus?','Rate your contribution towards extra curricular activities','Rate your technical competencies','What are your package expectations (LPA)','Internships Interests'])
```

```python
data
```

```python
data =data.rename(columns={"Your CIA % of last semester":"CIA","Your GPA of last semester":"GPA","Your maximum attendance % till last semester":"Attendance"})
```

```python
data.head(5)
```

```python
data.shape
```

```python
data.isnull().sum()
```

```python
data.dtypes
```

```python
data["CIA"] = data["CIA"].str.replace("%","")
data["Attendance"] = data["Attendance"].str.replace("%","")
```

```python
data['CIA'] = pd.to_numeric(data['CIA'], errors='coerce')
data['Attendance'] = pd.to_numeric(data['Attendance'], errors='coerce')
data['GPA'] = pd.to_numeric(data['GPA'], errors='coerce')
```

```python
print("Duplicates:", data.duplicated().sum())
```

```python
data = data[(data['CIA'] >= 0) & (data['CIA'] <= 100)]
```

```python
data = data[(data['Attendance'] >= 0) & (data['Attendance'] <= 100)]
```

```python
data = data[(data['GPA'] >= 0) & (data['GPA'] <= 10)]
```

```python
Q1 = data['CIA'].quantile(0.25)
Q3 = data['CIA'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

cia_outliers = data[(data['CIA'] < lower) | (data['CIA'] > upper)]

print("CIA Outliers:")
print(cia_outliers)
```

```python
Q1 = data['Attendance'].quantile(0.25)
Q3 = data['Attendance'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

attendance_outliers = data[(data['Attendance'] < lower) | (data['Attendance'] > upper)]

print("Attendance Outliers:")
print(attendance_outliers)
```

```python
Q1 = data['GPA'].quantile(0.25)
Q3 = data['GPA'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

gpa_outliers = data[(data['GPA'] < lower) | (data['GPA'] > upper)]

print("GPA Outliers:")
print(gpa_outliers)
```
