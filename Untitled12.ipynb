```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
student_performance = fetch_ucirepo(id=320)

# data (as pandas dataframes)
X = student_performance.data.features
y = student_performance.data.targets

# metadata
print(student_performance.metadata)

# variable information
print(student_performance.variables)
```

## Data Preprocessing

First, let's combine the features (`X`) and targets (`y`) into a single DataFrame for easier preprocessing.

```python
import pandas as pd

df = pd.concat([X, y], axis=1)
display(df.head())
```

### Handling Missing Values

Based on the dataset metadata, there should be no missing values, but it's good practice to verify.

```python
print("Number of missing values per column:")
print(df.isnull().sum())
```

### Encoding Categorical Variables

Many columns are categorical and need to be converted into a numerical format. I'll use one-hot encoding for this purpose.

```python
from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns (object type in pandas)
categorical_cols = df.select_dtypes(include=['object']).columns

# Apply one-hot encoding
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True) # drop_first to avoid multicollinearity

print("DataFrame after One-Hot Encoding:")
display(df_encoded.head())
```

### Feature Scaling

Numerical features, especially those with different ranges, often benefit from scaling before being fed into machine learning models. I'll use `StandardScaler` to standardize the numerical features.

```python
from sklearn.preprocessing import StandardScaler

X_preprocessed = df_encoded.drop(columns=['G1', 'G2', 'G3'], errors='ignore')
y_target = df_encoded[['G1', 'G2', 'G3']]

numerical_cols_to_scale = [col for col in X_preprocessed.columns if X_preprocessed[col].nunique() > 2 or X_preprocessed[col].dtype != 'uint8']

scaler = StandardScaler()
X_scaled = X_preprocessed.copy()
X_scaled[numerical_cols_to_scale] = scaler.fit_transform(X_preprocessed[numerical_cols_to_scale])

print("Scaled Features (X_scaled):")
display(X_scaled.head())

print("Target Variables (y_target):")
display(y_target.head())
```

## Splitting Data into Training and Testing Sets

Now, let's split our preprocessed features (`X_scaled`) and selected target variable (`y_G3`) into training and testing sets. This is crucial for evaluating the performance of any machine learning model on unseen data.

```python
from sklearn.model_selection import train_test_split



y_G3 = y_target['G3']


X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_G3, test_size=0.20, random_state=42)

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

## Implementing Linear Regression with Gradient Descent

Now, let's implement Linear Regression from scratch using the Gradient Descent algorithm. We'll define the hypothesis function, the cost function (Mean Squared Error), and the gradient descent update rules.

```python
import numpy as np

def hypothesis(X, weights, bias):
    """
    Linear regression hypothesis function: h(x) = X * w + b
    X: feature matrix
    weights: weight vector
    bias: bias term
    """
    return np.dot(X, weights) + bias

def cost_function(X, y, weights, bias):
    """
    Mean Squared Error cost function.
    X: feature matrix
    y: true target values
    weights: weight vector
    bias: bias term
    """
    m = len(y)
    predictions = hypothesis(X, weights, bias)
    sq_errors = (predictions - y)**2
    return (1 / (2 * m)) * np.sum(sq_errors)

def gradient_descent(X, y, weights, bias, learning_rate, num_iterations):
    """
    Performs gradient descent to optimize weights and bias.
    X: feature matrix
    y: true target values
    weights: initial weight vector
    bias: initial bias term
    learning_rate: step size for updates
    num_iterations: number of iterations to run gradient descent
    """
    m = len(y)
    cost_history = []

    for i in range(num_iterations):
        predictions = hypothesis(X, weights, bias)
        errors = predictions - y

        # Update weights and bias
        d_weights = (1 / m) * np.dot(X.T, errors)
        d_bias = (1 / m) * np.sum(errors)

        weights = weights - learning_rate * d_weights
        bias = bias - learning_rate * d_bias

        cost = cost_function(X, y, weights, bias)
        cost_history.append(cost)

        if i % (num_iterations // 10) == 0: # Print cost periodically
            print(f"Iteration {i}: Cost = {cost:.4f}")

    return weights, bias, cost_history
```

### Training the Linear Regression Model

Now, let's train our implemented Linear Regression model using the `X_train` and `y_train` data. We'll initialize random weights and bias, and then run gradient descent.

```python
import matplotlib.pyplot as plt

# Initialize parameters
n_features = X_train.shape[1]
initial_weights = np.zeros(n_features) # Initialize weights to zeros
initial_bias = 0 # Initialize bias to zero

# Set hyperparameters
learning_rate = 0.01
num_iterations = 1000

print("Starting Gradient Descent training...")
# Train the model
final_weights, final_bias, cost_history = gradient_descent(
    X_train.to_numpy(), # Convert DataFrame to NumPy array for dot product
    y_train.to_numpy(), # Convert Series to NumPy array
    initial_weights,
    initial_bias,
    learning_rate,
    num_iterations
)

print("\nTraining complete.")
print(f"Final Weights: {final_weights[:5]}... (first 5)")
print(f"Final Bias: {final_bias:.4f}")
print(f"Final Cost: {cost_history[-1]:.4f}")

# Plot the cost history
fig = plt.figure(figsize=(10, 6))
plt.plot(range(num_iterations), cost_history)
plt.title('Cost Function History (Gradient Descent)')
plt.xlabel('Number of Iterations')
plt.ylabel('Cost (MSE)')
plt.grid(True)
plt.show()
```

## Model Evaluation

Now that our Linear Regression model is trained, let's evaluate its performance on the test set (`X_test`, `y_test`). We'll make predictions and then calculate common regression metrics like Mean Squared Error (MSE) and R-squared ($R^2$).

```python
from sklearn.metrics import mean_squared_error, r2_score

# Make predictions on the test set
predictions_test = hypothesis(X_test.to_numpy(), final_weights, final_bias)

# Calculate Mean Squared Error (MSE)
mse = mean_squared_error(y_test, predictions_test)

# Calculate R-squared (R2 Score)
r2 = r2_score(y_test, predictions_test)

print(f"\nModel Evaluation on Test Set:")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R-squared (R2 Score): {r2:.4f}")

# Optionally, visualize predictions vs actual values
plt.figure(figsize=(10, 6))
plt.scatter(y_test, predictions_test, alpha=0.6)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2) # y=x line
plt.xlabel("Actual Grades (G3)")
plt.ylabel("Predicted Grades (G3)")
plt.title("Actual vs. Predicted Grades on Test Set")
plt.grid(True)
plt.show()
```

## Experimenting with a Different Learning Rate

Let's try a different `learning_rate` to observe its effect on model convergence and performance.

```python
print('--- Experimenting with a new learning rate ---')

# Set a new learning rate (e.g., higher or lower)
new_learning_rate = 0.1 # Let's try a higher learning rate

print(f"New learning rate: {new_learning_rate}")

# Re-initialize parameters for the new training run
initial_weights_new = np.zeros(n_features)
initial_bias_new = 0

print("Starting Gradient Descent training with new learning rate...")
# Train the model with the new learning rate
final_weights_new, final_bias_new, cost_history_new = gradient_descent(
    X_train.to_numpy(),
    y_train.to_numpy(),
    initial_weights_new,
    initial_bias_new,
    new_learning_rate,
    num_iterations
)

print("\nTraining complete with new learning rate.")
print(f"Final Weights (new learning rate): {final_weights_new[:5]}... (first 5)")
print(f"Final Bias (new learning rate): {final_bias_new:.4f}")
print(f"Final Cost (new learning rate): {cost_history_new[-1]:.4f}")

# Plot the cost history for the new learning rate
fig_new = plt.figure(figsize=(10, 6))
plt.plot(range(num_iterations), cost_history_new)
plt.title(f'Cost Function History (Gradient Descent) - Learning Rate: {new_learning_rate}')
plt.xlabel('Number of Iterations')
plt.ylabel('Cost (MSE)')
plt.grid(True)
plt.show()
```

```python
# Make predictions on the test set with the new model
predictions_test_new = hypothesis(X_test.to_numpy(), final_weights_new, final_bias_new)

# Calculate Mean Squared Error (MSE)
mse_new = mean_squared_error(y_test, predictions_test_new)

# Calculate R-squared (R2 Score)
r2_new = r2_score(y_test, predictions_test_new)

print(f"\nModel Evaluation on Test Set (New Learning Rate: {new_learning_rate}):")
print(f"Mean Squared Error (MSE): {mse_new:.4f}")
print(f"R-squared (R2 Score): {r2_new:.4f}")

# Visualize predictions vs actual values for the new model
plt.figure(figsize=(10, 6))
plt.scatter(y_test, predictions_test_new, alpha=0.6)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2) # y=x line
plt.xlabel("Actual Grades (G3)")
plt.ylabel("Predicted Grades (G3)")
plt.title(f"Actual vs. Predicted Grades on Test Set (Learning Rate: {new_learning_rate})")
plt.grid(True)
plt.show()
```

## Interpretation of Model Results

We have successfully implemented a Linear Regression model using Gradient Descent and experimented with two different learning rates (`0.01` and `0.1`). Let's summarize the convergence behavior and prediction performance.

### Convergence Behavior:

*   **Learning Rate 0.01 (Original):** The cost function decreased steadily over 1000 iterations, indicating a stable convergence towards a local minimum. The final training cost was approximately `3.2275`.
*   **Learning Rate 0.1 (New):** With a higher learning rate, the model converged much faster, reaching a stable cost significantly earlier than the `0.01` learning rate. The final training cost was slightly lower at `3.2238`.

    *The comparison plot of the cost histories clearly illustrated this: the higher learning rate's curve dropped sharply and flattened out quickly, while the lower learning rate's curve was a more gradual descent.* Both eventually reached similar minimum cost values, suggesting that the chosen `num_iterations` (1000) was sufficient for both to converge.

### Prediction Performance:

The evaluation on the test set using various metrics provides insight into the model's predictive power:

| Metric                   | Original Model (LR: 0.01) | New Model (LR: 0.1) |
| :----------------------- | :------------------------ | :------------------ |
| **Mean Absolute Error (MAE)** | `2.1457`                  | `2.1564`            |
| **Mean Squared Error (MSE)**  | `8.1062`                  | `8.1898`            |
| **Root Mean Squared Error (RMSE)** | `2.8471`                  | `2.8618`            |
| **R-squared (R² Score)** | `0.1687`                  | `0.1602`            |

**Key Observations:**

*   **MAE and RMSE:** These metrics indicate the average magnitude of the errors. An MAE of around `2.15` suggests that, on average, our model's predictions for the final grade (G3) are off by approximately 2.15 points. The RMSE, which penalizes larger errors more heavily, is around `2.85`.
*   **R-squared (R² Score):** The R-squared values for both models are very low (around `0.16` to `0.17`). This is the most significant finding. It means that our Linear Regression model, even after tuning the learning rate, explains only about 16-17% of the variance in the target variable ('G3'). The remaining 83-84% of the variance is unexplained by our current set of features and linear relationship.

**Conclusion:**

While the Gradient Descent algorithm successfully converged for both learning rates, and the higher learning rate did accelerate convergence, neither model demonstrates strong predictive performance for the 'G3' target variable. The low R-squared score is a strong indicator that a simple linear relationship, as modeled here, is likely not sufficient to capture the complexity of factors influencing student final grades in this dataset. This suggests that for better predictions, we might need to explore more complex models, advanced feature engineering (e.g., interaction terms, polynomial features), or consider that other uncaptured variables significantly impact student performance.
